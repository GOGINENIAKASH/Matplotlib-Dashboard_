```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(r"C:\Users\gogin\OneDrive\Documents\student_dataset.csv")
```

```python
# Create dashboard
fig, ax = plt.subplots(4, 2, figsize=(18, 20))


# 1. BAR CHART 

ax[0, 0].bar(
    df["name"],
    df["total_marks"],
    color="royalblue"
)

ax[0, 0].set_title("Student vs Total Marks")
ax[0, 0].set_xlabel("Student")
ax[0, 0].set_ylabel("Marks")
ax[0, 0].tick_params(axis="x", rotation=90)


# 2. PIE CHART 

placement = df["placement_status"].value_counts()

ax[0, 1].pie(
    placement.values,
    labels=placement.index,
    autopct="%1.1f%%",
    colors=["mediumseagreen", "tomato"]
)

ax[0, 1].set_title("Placement Status")



# 3. LINE CHART 

ax[1, 0].plot(
    df["name"],
    df["total_marks"],
    marker="o",
    color="purple",
    linewidth=2
)

ax[1, 0].set_title("Student Marks Trend")
ax[1, 0].set_xlabel("Student")
ax[1, 0].set_ylabel("Marks")
ax[1, 0].tick_params(axis="x", rotation=90)
ax[1, 0].grid()



# 4. HISTOGRAM 

ax[1, 1].hist(
    df["total_marks"],
    bins=8,
    color="orange",
    edgecolor="black"
)

ax[1, 1].set_title("Distribution of Total Marks")
ax[1, 1].set_xlabel("Marks")
ax[1, 1].set_ylabel("Number of Students")



# 5. SCATTER PLOT 

ax[2, 0].scatter(
    df["study_hours"],
    df["total_marks"],
    color="teal",
    s=70
)

ax[2, 0].set_title("Study Hours vs Marks")
ax[2, 0].set_xlabel("Study Hours")
ax[2, 0].set_ylabel("Total Marks")
ax[2, 0].grid()



# 6. BOXPLOT 

ax[2, 1].boxplot(
    df["total_marks"],
    patch_artist=True,
    boxprops=dict(facecolor="lightcoral")
)

ax[2, 1].set_title("Total Marks Boxplot")
ax[2, 1].set_ylabel("Marks")


# 7. VIOLIN PLOT 

parts = ax[3, 0].violinplot(
    df["total_marks"],
    showmeans=True,
    showmedians=True
)

for body in parts["bodies"]:
    body.set_facecolor("mediumseagreen")

ax[3, 0].set_title("Total Marks Violin Plot")
ax[3, 0].set_ylabel("Marks")


# 8. SCATTER PLOT 

ax[3, 1].scatter(
    df["attendance"],
    df["total_marks"],
    color="magenta",
    s=70
)

ax[3, 1].set_title("Attendance vs Marks")
ax[3, 1].set_xlabel("Attendance (%)")
ax[3, 1].set_ylabel("Total Marks")
ax[3, 1].grid()



fig.suptitle(
    "STUDENT PERFORMANCE DASHBOARD",
    fontsize=22,
    fontweight="bold"
)
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
