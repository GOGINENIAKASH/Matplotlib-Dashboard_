```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(r"C:\Users\gogin\OneDrive\Documents\house_prices_500.csv")
```

```python

fig, ax = plt.subplots(2, 2, figsize=(15, 10))

# 1. Area vs Price

ax[0, 0].scatter(
    df["Area"],
    df["Price"],
    color="royalblue",
    alpha=0.7
)

ax[0, 0].set_title("Area vs House Price")
ax[0, 0].set_xlabel("Area (sq ft)")
ax[0, 0].set_ylabel("Price")
ax[0, 0].grid(alpha=0.3)


# 2. Bedrooms vs Price

bedrooms = sorted(df["Bedrooms"].unique())

price_by_bedroom = [
    df[df["Bedrooms"] == b]["Price"]
    for b in bedrooms
]

ax[0, 1].boxplot(
    price_by_bedroom,
    tick_labels=bedrooms,
    patch_artist=True
)

ax[0, 1].set_title("Price Distribution by Bedrooms")
ax[0, 1].set_xlabel("Number of Bedrooms")
ax[0, 1].set_ylabel("Price")
ax[0, 1].grid(axis="y", alpha=0.3)



# 3. Location vs Average Price

location_price = (
    df.groupby("Location")["Price"]
    .mean()
    .sort_values()
)

ax[1, 0].bar(
    location_price.index,
    location_price.values,
    color="seagreen"
)

ax[1, 0].set_title("Average House Price by Location")
ax[1, 0].set_xlabel("Location")
ax[1, 0].set_ylabel("Average Price")
ax[1, 0].tick_params(axis="x", rotation=30)



# 4. Age vs Price

ax[1, 1].scatter(
    df["Age"],
    df["Price"],
    color="darkorange",
    alpha=0.7
)

ax[1, 1].set_title("House Age vs Price")
ax[1, 1].set_xlabel("House Age (Years)")
ax[1, 1].set_ylabel("Price")
ax[1, 1].grid(alpha=0.3)


# Dashboard Title

fig.suptitle(
    "HOUSE PRICE ANALYSIS DASHBOARD",
    fontsize=20,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
```
